```python
import os

def add_py_files_to_single_file(directory, output_file):
    with open(output_file, 'w') as outfile:
        for root, _, files in os.walk(directory):
            for file in files:
                if file.endswith('.py') and file != output_file:
                    file_path = os.path.join(root, file)
                    with open(file_path, 'r') as infile:
                        outfile.write(f"# Content from: {file_path}\n")
                        outfile.write(infile.read())
                        outfile.write("\n\n")

add_py_files_to_single_file('src', 'whole_code_for_interpreter.py')
```
