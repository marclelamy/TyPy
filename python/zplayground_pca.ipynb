```python
import sqlite3
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.io as pio
from src.score import Score
import os 
import toml
from tabulate import tabulate
import pandas as pd
import sqlite3
from sklearn.decomposition import PCA

current_dir = os.getcwd()
con = sqlite3.connect("data/main_database.db")
# pd.set_option('display.float_format', lambda x: '%.3f' % x)
pd.set_option('display.max_columns', None)
# connew = sqlite3.connect("data/main_database_new.db")

# Setting pandas options
pd.options.display.max_columns = None
pd.options.display.float_format = '{:,}'.format

# Setting plotly as dark
pio.templates.default = "plotly_dark"
```

```python
df = pd.read_sql_query('select * from games_summary', con)
df = df[['game_id', 'accuracy', 'wpm', 'keys_to_press']].copy().dropna().reset_index(drop=True)
df
```

```python
import pandas as pd
from sklearn.decomposition import PCA

# Load the original DataFrame with three variables
# df = pd.read_csv('data.csv')

# Create a PCA object with one component
pca = PCA(n_components=1)

# Fit the PCA model to the data and transform the data to get the principal component
principal_component = pca.fit_transform(df)

# Convert the principal component to a DataFrame and add a column name
pc_df = pd.DataFrame(principal_component, columns=['PC1'])
df['pca'] = principal_component
```

```python
px.histogram(df, x='pca', nbins=100, title='PCA')
```

```python
fig = px.scatter_3d(df,
              x='accuracy',
              y='wpm',
              z='keys_to_press',
              color='pca',)


fig.write_html("zplayground_pca.html")
```
